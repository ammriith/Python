```python
#EXCHANGE STRING
s=str(input("enter a string:"))
print("before swapping:",s)
a=s[0]
b=s[-1]
c=b+s[1:-1]+a
print("after swapping:",c)
```

```python
#A STRING FROM TWO STRING AND SWAP FIRST TWO CHARACTERS
a1=str(input("Enter the first string:"))
a2=str(input("Enter the second string:"))
print("result string:",a1,a2)
x=a1[0:2]
z=a2[0:2]
a3=z+a1[2:]
a4=x+a2[2:]
print("new string:",a3,a4)
```

```python
#REPLACE THE LETTER WITH $
s=str(input("enter the string:"))
a=s[0]
b=s[1:]
st=b.replace(a,"$")
ns=s[0]+st
print("new string:",ns)
```

```python
#COLOR SET
f=set(input("Enter color for set f(comma-seperated):").split(","))
s=set(input("Enter color for set s(comma-seperated):").split(","))
difference_f_s=f-s
print(f"difference of {f} and {s}:{difference_f_s}")
```

```python
#ascending and descendimg order
student_names=input("Enter 3 student names seperated by comma: ").split(',')
students=dict(enumerate(student_names))
print("students sorted in ascending order:",dict(sorted(students.items())))
print("students sorted in descending order:",dict(sorted(students.items(),reverse=True)))
      
      

```

```python
#EVEN NUMBERS
num=range(0,25,2)
print("Even numbers 0 to 25:",list(num))
print("sum of even numbers 0 to 25:",sum(num))
```

```python
#MERGED DICTIONARY
dic1=input("Enter the name of 3 students name seperated by comma:").split(",")
m={1:dic1[0], 2:dic1[1], 3:dic1[2]}
print("first dictionary:",m)
dic2=input("Enter the name of 2 students name seperated by comma:").split(",")
n={4:dic2[0], 5:dic2[1]}
print("second dictionary:",n)
m.update(n)
print("merged dictionary:",m)
```
