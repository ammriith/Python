```python
#OCCURANCE OF 'a' IN THE LIST
names=input("Enter the list of first name:").split()
count = 0
for name in names:
    count += name.count('a')
print(f"The letter 'a' appears {count} times in the list of names.")
```

```python
#LENGTH,SUM,COMMON VALUES
list1 =list(map(int,input("Enter the first list of integers : ").split()))
list2 = list(map(int,input("Enter the second list of integers : ").split()))
if len(list1) == len(list2):
    print("The length lists are of the same.")
else:
    print("The length lists are of the not same.")
if sum(list1) == sum(list2):
    print("The sum of list are same.")
else:
    print("The sum of list are not same.")
    common = set(list1) & set(list2) 
if common:
    print(f"Common values between the lists: {common}")
else:
    print("There are no common values between the lists.")


```

```python
#LIST COMPREHENSIONS
n=list(map(int,input("Enter a list of integer:").split(',')))
word=input("Enter a word:")
positive_number=[num for num in n if num > 0]
print("positive number:",positive_number)
square=[i**2 for i in n]
print("square of N numbers :",square)
vowels=[char for char in word if char in'aeiouAEIOU']
print("vowels in a word:",vowels)
ordinals=[ord(char) for char in word]
print("ordinal values of word:",ordinals)
```
