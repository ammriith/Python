```python
#GCD
import math


n1 = int(input("Enter the first number: "))
n2 = int(input("Enter the second number: "))


gcd = math.gcd(n1, n2)

print("GCD of", n1, "and", n2, "is:", gcd)
```

```python
#REMOVING EVEN NUMBERS
n = list(map(int,input("Enter the numbers : ").split()))
odd_numbers = [num for num in n if num % 2 != 0]
print("List with even numbers removed:", odd_numbers)
```

```python
#SUM OF ALL ITEMS
n = list(map(int, input("Enter the numbers: ").split()))
s = sum(n)
print("Sum of all items in the list:",s)
```

```python
#CHARACTER FREQUENCY
from collections import Counter
s = input("Enter a string: ")
count = Counter(s)
for char, count_value in count.items():
    print(f"'{char}': {count_value}")

```

```python
#ADD ING
s = input("Enter a string: ")
if s.endswith('ing'):
    s += 'ly'
else:
    s += 'ing'
print("new string:", s)
```

```python
#PERFECT SQUARE
import math

start = int(input("Enter the starting index: "))
end = int(input("Enter the ending index: "))

start = max(1000, start)
end = min(9999, end)

result = []

for i in range(start, end + 1):
    if int(math.sqrt(i)) ** 2 == i:
        if all(int(digit) % 2 == 0 for digit in str(i)):
            result.append(i)

print("Four digit even perfect square numbers:", result)
```
