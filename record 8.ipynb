```python
#COMPARE TWO RECTANGLES 

class Rectangle:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth
    
    def area(self):
        return self.length * self.breadth
    
    def perimeter(self):
        return 2 * (self.length + self.breadth)
    
    
    
   
    def __lt__(self, other):
        return self.area() < other.area()
    
    def __eq__(self, other):
        return self.area() == other.area()


l = int(input("Enter the length of the rectangle 1: "))
b = int(input("Enter the breadth of the rectangle 1: "))
rect1 = Rectangle(l, b)

l = int(input("Enter the length of the rectangle 2: "))
b = int(input("Enter the breadth of the rectangle 2: "))
rect2 = Rectangle(l, b)


print(f"Rectangle 1 area = {rect1.area()}, perimeter = {rect1.perimeter()}")
print(f"Rectangle 2 area = {rect2.area()}, perimeter = {rect2.perimeter()}")


if rect1 < rect2:
    print("Area of object 2 is greater than object 1")
elif rect1 == rect2:
    print("Area of object 1 is equal to object 2")
else:
    print("Area of object 1 is greater than object 2")
```

```python
#PUBLISHER
class publisher:
    def __init__(self,pname):
        self.pubname=pname
    def display1(self):
        print("Book details ")
        print("Publisher name=",self.pubname)
class book(publisher):
    def __init__(self,pname,bname,author):
        self.title=bname
        self.auth=author
        publisher.__init__(self,pname)
    def display2(self):
        print("Title=",self.title)
        print("Author",self.auth)

class python(book):
    def __init__(self,pname,title,au,price,np):
        self.rate=price
        self.nop=np
        book.__init__(self,pname,title,au)
    def display3(self):
        print("Rate=",self.rate)
        print("No. of pages=", self.nop)
p=input("Enter publisher name: ")
b=input("Enter book name: ")
a=input("Enter author name: ")
r=int(input("Enter rate of book: "))
n=int(input("Enter no. of pages: "))
a=python(p,b,a,r,n)
a.display1()
a.display2()
a.display3()
```
