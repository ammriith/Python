```python
#LEAP YEAR
def leap_year(year):
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        return True
    else:
        return False
c = int(input("Enter the current year: "))
f = int(input("Enter the final year: "))
print(f"Future leap years from {c} to {f}:")
for year in range(c + 1, f + 1):
    if leap_year(year):
        print("Leap year:", year)

```

```python
#FACTORIAL
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
n=int(input("Enter the number:"))
print(f"The factorial of {n}: {factorial(n)}")
```

```python
#FIBONACCI 
def fibonacci(n):
    fib = [0, 1]
    for i in range(2, n):
        fib.append(fib[i - 1] + fib[i - 2])
    return fib[:n]  

n = int(input("Enter a number: "))
print(f"Fibonacci series up to {n} terms: {fibonacci(n)}")
```

```python
#LENGTH OF LONGEST WORD
def length(word_list): 
    long_word = max(word_list, key=len) 
    return len(long_word)  

w = input("Enter a list of words: ").split()  
l = length(w)  
print("Length of the longest word:", l)  

```

```python
#PATTERN
def print_upper_half(n):
    for i in range(1,n+1):
        for j in range(i):
            print("*",end=" ")
        print()
def print_lower_half(n):
    for i in range(n-1,0,-1):
        for j in range(i):
            print("*",end=" ")
        print()
def print_pattern(n):
    print_upper_half(n)
    print_lower_half(n)
n=5
print_pattern(n)
```

```python
#PYRAMID
def print_pyramid(N):
    for i in range(1,N + 1):
        row=[]
        for j in range(1,i + 1):
            row.append(i * j)
        print("\t".join(map(str,row)))
N=int(input("Enter the number of steps(N):"))
print_pyramid(N)
```

```python
#FACTORS
def generate_factors(num):
    factors=[]
    for i in range(1,num+1):
        if num % i == 0:
            factors.append(i)
    return factors
num=int(input("Enter a number :"))
factors=generate_factors(num)
print(f"Factors of {num} are:{factors}")
```

```python
#AREA USING LAMBDA
def area_of_square(side):
    return side ** 2
def area_of_rectangle(length,width):
    return length * width
def area_of_triangle(base,height):
    return 0.5 * base * height

a=int(input("Enter the side:"))
b=int(input("Enter the length:"))
c=int(input("Enter the width:"))
e=int(input("Enter the base:"))
f=int(input("Enter the height:"))
square_area=area_of_square(a)
rectangle_area=area_of_rectangle(b,c)
triangle_area=area_of_triangle(e,f)
print(f"Area of Square is :{square_area}")
print(f"Area of Rectangle is :{rectangle_area}")
print(f"Area of Triangle is :{triangle_area}")
```
